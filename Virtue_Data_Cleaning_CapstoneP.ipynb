```python
#Importing Libraries
import pandas as pd
import numpy as np
```

```python
#Loading the Dataset into a pandas DataFrame called df
df = pd.read_csv('/content/diabetic risk data for grp 9.csv')
```

```python
# Remove the income_bracket column from the dataset

df.drop(columns=['income_bracket'], inplace=True)
```

```python
#Checking Missing Values
df.isnull().sum()
```

```python
#Giving an Overview of the Dataset, including data types and non_missing values
df.info()
```

```python
#Calculating the Mode for Smoking-status

df['smoking_status'].mode([0])
```

```python
# Filling the number of missing smoking_status values with the most common category (Never)

df['smoking_status']= df['smoking_status'].fillna('Never')
```

```python
#Since a large proportion of alcohol-consumption values are missing, we use "Unknown"

df['alcohol_consumption'] = df['alcohol_consumption'].fillna('unknown')
```
