```python
#Importing Libraries
import pandas as pd
import numpy as np
```

```python
#Loading the Dataset into a pandas DataFrame called df
df = pd.read_csv('/content/diabetic risk data for grp 9.csv')
```

```python
#Checking Missing Values
df.isnull().sum()
```

```python
#Giving an Overview of the Dataset, including data types and non_missing values
df.info()
```

```python
#Checking Income-bracket
df['income_bracket'].value_counts(dropna=False)
```

```python
#Mode Calculation for Income-bracket
df['income_bracket'].mode()[0]
```

```python
#Filling the missing Income values with the Mode
df['income_bracket'] = df['income_bracket'].fillna(df['income_bracket'].mode()[0])
```

```python
#Counting the number of observations in each income bracket category
df['income_bracket'].value_counts()
```

```python
#Checking Smoking Status
df['smoking_status'].value_counts(dropna=False)
```

```python
#Mode Calculation for Smoking Status
df['smoking_status'].mode()[0]
```

```python
#Filling the missing Smoking Status value with the mode
df['smoking_status'] = df['smoking_status'].fillna(df['smoking_status'].mode()[0])
```

```python
#Counting the number of observations in each Smoking status category
df['smoking_status'].value_counts()
```

```python
#Checking Alcohol Consumption
df['alcohol_consumption'].value_counts(dropna=False)
```

```python
#Mode Calculation for Alcohol Consumption
df['alcohol_consumption'].mode()[0]
```

```python
#Filling the missing Alcohol Consumption values with the mode
df['alcohol_consumption'] = df['alcohol_consumption'].fillna(df['alcohol_consumption'].mode()[0])
```

```python
#Counting the number of observations in each Alcohol Consumption category
df['alcohol_consumption'].value_counts()
```

```python
#Summarising Numerical variables for unususally high or low values
df.describe()
```
